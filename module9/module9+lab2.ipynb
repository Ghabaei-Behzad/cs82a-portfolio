Migration: Moving from Google Colab to AWS SageMaker Studio Lab

These are the required steps, reusable actions, and risks associated with moving our lab's data science workflow from Google Colab to AWS SageMaker Studio Lab. Which is free, and cloud based.

1. What Transfers Are Untouched during a Migration?

Both platforms are built as open-source Jupyter Notebooks.  Each project will migrate without any code modifications.
The .ipynb files and notebook structure, such as the text cells, markdown formatting, and cell execution will transfer exactly.
The Python Code such as, control loops, and data processing are the same.
The Libraries that work for python code like pandas, numpy, scikit-learn, and matplotlib will run the same.

2. What Needs Redoing

The Cloud notebook environments handles authentication, file systems, and copies and downloads differently.

Regarding Accounts and Authentication, and the Platform Access, you must register for an AWS SageMaker Studio Lab account using your specific, school permission, working email. This operates independently of standard AWS cloud credentials.  The API Credentials, and any access keys use SageMaker's terminal.

Regarding the File Storage And Data Access, you remove Google code uploading a dataset, on SageMaker.  The Storage for SageMaker Studio Lab provides a permanent 15 GB block storage  across sessions. You must upload files directly into this directory using the JupyterLab file interface or fetch data with wget or curl.


Regarding Packages Installment,  unlike Colab, where !pip install must be rerun every time the notebook disconnects, SageMaker Studio Lab saves your environment.  Using a Conda Environment, instead of running inline pip commands in your notebook, you will use the integrated Jupyter bash terminal to build an environment. Here is an example code snippet found on the SageMaker website.

conda create --name lab_env python=3.10
conda activate lab_env
pip install -r requirements.txt

3. The Risks of a Migration, which is most critical is the temporary loss of data access limits and potential compute downtime.
Unlike Google Colab's  easily upgradeable runtime, SageMaker Studio Lab enforces a strict 15 GB storage max and restricts compute sessions to 8 hours for CPU or 4 hours for GPU per day. If your dataset exceeds 15 GB, or if your model training sequences require long, runtimes, the notebook will halt, resulting in lost training checkpoints.
