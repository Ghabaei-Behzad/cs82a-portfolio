```python
import sqlite3
import pandas as pd

con = sqlite3.connect("database.db")
```

```python
create_table_sql = """
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    city TEXT,
    credit_limit REAL
);
"""
```

```python
insert_rows_sql = """
INSERT OR REPLACE INTO customers (id, name, city, credit_limit) VALUES
(1, 'Ali Husseini', 'New York', 5000.00),
(2, 'Bobby Jones', 'Los Angeles', 3500.50),
(3, 'John Brown', 'Chicago', 2500.00),
(4, 'Billy Prince', 'Boston', 7500.00),
(5, 'Eve Kim', 'Austin', 4200.00),
(6, 'Linda James', 'Chicago', 1500.75),
(7, 'Cathy Clark', 'Miami', 6000.00),
(8, 'Hilda Bosche', 'Seattle', 3100.00),
(9, 'Cynthia Bush', 'Miami', 3200.00),
(10, 'Cathy Jones', 'Miami', 3400.00),
(11, 'Annie Smith', 'Miami', 2200.00),
(12, 'Casey Blanc', 'Miami', 1200.00),
(13, 'Catalina Caballo', 'Miami', 1200.00);
"""
```

#### Here the "text_search_using_LIKE" variable returns all the records in the "customers" table that were from the "city" of "Miami" and have letters like "Ca" in their "name" which are ordered by the "credit_limit" they carry on their credit card. The table should return only Miami residents that have Ca in their name such as Catalina, Casey, and Cathy, but not Cynthia or Annie. SQLBolt.com can help to learn the use of the LIKE keyword. (sqlbolt.com | "SQL Lesson 3: Queries with constraints (Pt. 2)" | https://sqlbolt.com/lesson/select_queries_with_constraints_pt_2 )

```python
 text_search_using_LIKE = """
 SELECT * 
 FROM customers
 WHERE city == "Miami"
 AND name LIKE "%Ca%"
 ORDER BY credit_limit;
 
 """
```

```python
df_chart = pd.read_sql_query("SELECT * FROM customers;", con)
df_chart
```

#### LIKE keyword returns these 4 records, with "Ca" in their name, (using "% %" quotes and the percent sign.)  WHERE keyword returns these records who are people who reside in Miami. The credit limit is sorted, least to greatest with the key words ORDER BY. WHERE uses the AND keyword with LIKE.  The list returns: Casey Blanc, Catalina Caballo, Cathy Jones, and Cathy Clark (Cathy Clark has the highest credit limit)

```python
df_query = pd.read_sql_query( text_search_using_LIKE, con )
df_query
```

#### % is a wildcard character used for any string of zero or more charcters, for example WHERE city LIKE '%Los Angeles%' finds all cities with Los Angeles anywhere in the city column.

```python
sql_ex = """
SELECT * FROM customers
WHERE city
LIKE "%Los Angeles%";
""" 
done_ex = pd.read_sql_query(sql_ex, con)
done_ex
```

```python
con.execute(create_table_sql)
con.execute(insert_rows_sql)
con.execute(text_search_using_LIKE)
```
