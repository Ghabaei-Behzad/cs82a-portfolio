## Problem 7.1: What drives a restaurant’s overall rating? From ambience, food, and service ratings. The python code model.coef_ returns one slope per feature, rank them.  Which factor should a struggling restaurant fix first?

#### Mathematically the Multiple Linear Regression has a goal of modeling the relationship of the outcome variable Y and the predictor variables of X.  The y=mx+b formula shows: Overall Rating = B0 + B1(Ambience) + B2(Food) + B3(Service)  The Overall rating is Y. The outcome and dependent variable we are trying to predict.
#### Ambience, Food, and Service are predictors and independent variables representing X3 X2 X1
#### B0 is the intercept or the constant. This is the value of the Overall rating when all the predictors are 0.
#### B1,B2,B3 are the coefficients. These have an impact. For example, B1 is the average change in the Overall Rating for every unit changed in Ambience, when Food and Service are the same.
#### There is also the error term that is unexplainable, such as: Overall rating = B0 + B1(Ambience) + B2(Food) + B3(Service) + error

## Solving by Python, 
#### we want to find the coefficients with libraries like scikit-learn or statsmodels, to get the linear regression model.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
df = pd.read_csv("ratings.tsv", sep="\t")  
```

```python
print(df.shape)
print(df.info())
print(df.describe())
print(df.head(15))
print(df.columns)
```

#### Separate into predictors (X) and outcome (y) and Initialize and fit the Linear Regression model, then print the coefficients of each three columns and intercept.

```python
X = df[['ambience', 'food', 'service']]
y = df['rating']

model = LinearRegression()
model.fit(X, y)

print("--- Linear Regression Model Results ---")
print(f"Intercept (Beta 0): {model.intercept_:.4f}")
print(f"Ambience Coefficient (Beta 1): {model.coef_[0]:.4f}")
print(f"Food Coefficient (Beta 2): {model.coef_[1]:.4f}")
print(f"Service Coefficient (Beta 3): {model.coef_[2]:.4f}")
```

#### The formula y = mx + b, used for Overall rating says that:
#### Overall rating = Intercept + Ambience + Food + Service
#### The optimal multiple linear regression model for the dataset is:
#### Rating = -15.8713 + 0.7338(Food) + 0.1338(Ambience) + 0.3240(Service)

#### The coefficients measure how much the overall rating increases for every 1-point increase in a specific category.
#### Rating (78) = -15.8713 + 0.7338 * 85(Food) + 0.3240 * 89(Service) + 0.1338 * 82(Ambience)
#### 86.3 = -15.87 + 62.373 + 28.836 + 10.9716
#### 86 - 78 = 8 point error difference.

#### Rating (Y) depends on the three predictors. Food (X1) Ambience(X2) Service(X3)
#### Y = B0 + B1 X1 + B2 X2 + B3 X3 
#### B0 is the intercept or constant, and B1 B2,B3 are coefficients that impact the formula.


## What drives a restaurant’s overall rating?

### The business recommendation I would make is that a struggling restaurant must prioritize food quality first, because it holds more than double the statistical influence more than service and over five times the influence more than ambience.  According to the largest coefficient at 0.7338, Food quality is the most influential and weight carrying aspect of a restaurant's overall rating.  Based on the linear regression model trained on this dataset, the features that impact a customer's satisfaction are shown as coefficients, and this three-feature regression model which fits the data, assists the business recommendation which follows the mathematics. The coefficients are obtained from ordinary least squares (OLS) from the linear regression algorithm. The features, Food, Ambience, and Service are matched as the slope: The Intercept slope is B0: -15.8713 , the Food slope is B1: +0.7338, and the Ambience Slope is B2: +0.1338, lastly the Service Slope is B3: +0.3240. The formula says: **Rating = -.1587 + 0.7338 (Food) + 0.1338 (Ambience) + 0.3240 (Service).**  The Ranked Order is **Food as the Largest Coefficient (most influential), Service as the Median Coefficient, and Ambience as the Smallest Coefficient.**  In linear regression, the size of the coefficient represents the sensitivity of the outcome variable to a 1-unit change in that specific predictor. Since Food (0.7338) is the highest, a struggling restaurant will get the maximum possible increase in overall ratings by improving the restaurant kitchen and food. For example, raising the the Food score by 10 points gets a large point increase in the overall rating. However, a 10-point improvement in Ambience gets a small point increase. Thus, focusing on the food served from the kitchen, results in an efficient use of investments.
