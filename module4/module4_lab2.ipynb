## Step 4: The volume variance

#### The variance is used to measure how spread the data points are. First we pick a center distribution like the mean and measure distance from each data point. The two types of variance are variance of a population, and variance of a sample. The variance of volume is very large at 115518.35

```python
df["volume"].var()
```
