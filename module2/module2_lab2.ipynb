### Do it with a **While Loop**

```python
# Problem 4.2
# when multiplying x * y, assignment operators must be done with result += x
# result = result + x
first_number = 7
second_number = 6
# initialize these to 0, result uses the augmented assignment operator
def while_multiply(first_num, second_num):
    count = 0
    result = 0
    while(count < second_num):      # how many times should we add the first number to itself? Until count reaches the second number.
        count += 1                  # count the iterations
        result = result + first_num # accumulate the sum
    return result

answer = while_multiply(first_number, second_number)
print("answer is:", answer)
print(first_number, "*", second_number,"=", answer)
```

### **How does the while loop terminate?** The while loop always iterates if the condition is true. Remeber to set a count and **increment the count ( or index) or it will loop infinitely.** In other words "A while loop needs its counter updated inside the loop, or it runs forever."  Set a limit with a comparison operator.  Comparison operators return True or False depending on the comparison. (https://www.w3schools.com/python/python_operators_comparison.asp)

### Do it with a **For Loop**

```python
# initialize these to 0, result uses the augmented assignment operator result += first_num
# To loop through a set of code a specified number of times, we can use the range() function,
# The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number. (https://www.w3schools.com/python/python_for_loops.asp)
def for_multiply(first_num, second_num):
    count = 0
    result = 0
    for count in range(second_num):    # range begins from 0 to parameter, "_" can be used instead of count as a placeholder. 
        # count += 1                   # range already increments count for us.   
        result = result + first_num    # accumulate the sum
    return result

answer_2 = for_multiply(first_number, second_number)
print("answer is:", answer_2)
print(first_number,"*", second_number,"=", answer_2)
```

### This **for loop** uses the range() function to increment up to the parameter. The keyword "in" links to the sequence you want to iterate over.  The statement is executed each iteration.
