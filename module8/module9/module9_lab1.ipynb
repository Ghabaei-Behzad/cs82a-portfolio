```python
def convert_lbs_to_kilograms(v):
  return v * 0.453592

lbs_a = 10
print(lbs_a, "lbs: ", convert_lbs_to_kilograms(lbs_a), "kg")
lbs_b = 20
print(lbs_b, "lbs: ", convert_lbs_to_kilograms(lbs_b), "kg")
lbs_c = 30
print(lbs_c, "lbs: ", convert_lbs_to_kilograms(lbs_c), "kg")
lbs_d = 150
print(lbs_d, "lbs: ", convert_lbs_to_kilograms(lbs_d), "kg")

v = convert_lbs_to_kilograms(lbs_d)
def kilograms_to_milligrams(v):
  return v * 1000000

print(v, "kg: ", kilograms_to_milligrams(v), "mg")

milligrams_out = kilograms_to_milligrams(v)

def milligrams_to_kilograms(milligrams_out):
  v = milligrams_out
  return v * 0.000001

print(milligrams_out,"mg: ", milligrams_to_kilograms(milligrams_out), "kg")



```
