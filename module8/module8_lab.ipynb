## Step 4: Run the AI’s code

You should see: a missing-data summary, a species comparison, and at least one chart.

```python
# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# (1) Summarize missing data
# We compute the count of missing values per column to understand completeness
missing_summary = df.isnull().sum()
print("--- (1) Missing Data Summary ---")
print(missing_summary)

# (2) Compare flipper_length_mm across species by computing the group means
# We group the dataset by species and find the mean flipper length
species_flipper_means = df.groupby('species')['flipper_length_mm'].mean()
print("\n--- (2) Group Means for Flipper Length (mm) ---")
print(species_flipper_means)

# (3) Produce one labeled chart showing this distribution
# We use a boxplot to display the distribution of flipper_length_mm across the different species
print("\n--- (3) Display the distribution of flipper_length_mm across the different species  ---")
plt.figure(figsize=(8, 6))
sns.boxplot(x='species', y='flipper_length_mm', hue='species', data=df, palette='Set2', legend=False)
plt.title('Distribution of Flipper Length across Penguin Species')
plt.xlabel('Species')
plt.ylabel('Flipper Length (mm)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

AI prompt: (3) Display the distribution of flipper_length_mm across the different species using a bar plot instead of boxplot.

```python
# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Compute mean flipper length per species
species_means = df.groupby('species')['flipper_length_mm'].mean().reset_index()

# Create a bar plot representing the means
plt.figure(figsize=(8, 6))
sns.barplot(x='species', y='flipper_length_mm', hue='species', data=species_means, palette='Set2', legend=False)
plt.title('Average Flipper Length by Penguin Species')
plt.xlabel('Species')
plt.ylabel('Mean Flipper Length (mm)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
