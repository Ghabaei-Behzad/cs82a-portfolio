## Step 2: Write your first observations

#### Looking at the data, many concerns arise about the precision of the data. First, using floats may not be a good idea, because a death is either true or false. Using integers will reflect the true nature of deaths.  Integers will eliminate noise and save storage.  Urban populations should not exceed 100% and if they are integers, it works best for clarity, instead of decimal point values.  An assault could be used as an integer, which would be best, so that decimal assaults don't cause confusion.   Lets change the data types of each of these to integers.
#### Next, lets keep the urban population percentage limited and cut off at 100. Looking at impossible and extreme values of each entry is done with df.head(50) which can list them all. 
#### Assault is missing a value, that should be filled.

## Step 3: Count the missing values

````python
# apply tutorial 1 to the whole dataframe
df.isna().sum() #  1 missing value under assault
````

## Step 4: Impute the missing value, dropping rows loses valuable data.

````python
df.shape # run this before and after filling missing values.
````

````python
# Fill the missing prices using the Tutorial 2 pattern
# looking for: Impossible Values, Suspicious Outliers, Missing Values, Floating Point Noise
fill_value = df["Assault"].median() # The 50% median of "Assault" when running df.describe() is 159.00 
# whether mean or median fits your lab data better 
df["Assault"] = df["Assault"].fillna(fill_value)
print(df.head(20))
````

````python
df.shape
````

````python
df.isnull().sum()
````

#### Assault no longer has a null value.  The 50 percentile replaced the NaN.

## Step 5: Remove duplicate rows

````python
print(df.duplicated().sum()) # counts fully repeated rows
df = df.drop_duplicates()    # removes them. 
````

#### We have succesfully pre-processed the data.  First adressing all the missing values.  Second, looking for outliers and noisy data to smooth, with data type conversion and min, max, adjustments, and with percentage cut off at 100.  Lastly, lets establish a relation between urban population category and crime type.  
- Small < 50% 
- medium < 60%
- large < 70%
- extra-large >= 70%

#### We will check pandas.pydata.org for suggestions on how to categorize datasets. "User Guide" > "Table Visualization" > here is the snippet of a function with the if-elif-else condition:
```
[2]: 
def rain_condition(v):
    if v < 1.75:
        return "Dry"
    elif v < 2.75:
        return "Rain"
    return "Heavy Rain"

styler.format(rain_condition)
```
#### ( https://pandas.pydata.org/docs/user_guide/style.html )

````python
def UrbanPop_condition(v):
    if v < 50:
        return "Small"
    elif v < 60:
        return "Medium"
    elif v < 70:
        return "Large"
    else:
        return "Extra-Large"
# the apply function was needed here
df['UrbanPop_Category'] = df["UrbanPop"].apply(UrbanPop_condition)
df.head(60)
````
