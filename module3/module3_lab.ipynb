## Step 4: Impute the missing prices

#### We'll be using the median for the missing values of price.  This is much better than the average because it represents the 50 percentile of the data.  This value is 37.5. Since the data only has 12 missing values, it might be better to put them in, one by one, by hand, instead of using a set price.  This is because the merchandise ranges from notebooks and pen sets, to monitors and chargers, which are considerably more expensive.  However, we will use the median instead of the mean.

```python
df.shape # always run shape before and after imputing missing prices.
```

```python
# Fill the missing prices using the Tutorial 2 pattern
# looking for: Impossible Values, Suspicious Outliers, Missing Values, Floating Point Noise
fill_value = df["price"].median() # The 50% median of price when running df.describe() is 37.53
# whether mean or median fits your lab data better 
df["price"] = df["price"].fillna(fill_value)
print(df.head(10))
```

```python
df.shape
```

#### Price no longer has a null value, as shown below, 0 missing prices afterwards

```python
df.isna().sum() # df.isnull().sum() - This makes price ---- 0
```

## Step 5: Remove duplicate rows

#### According to duplicated().sum() functions there are 8 repeated values which can skew our data.  By removing these 8, the data can become more accurate, and the average will not be pulled away from the actual.  These extra 8 duplicates can be the cause of inaccurate results and suspicious outliers.

```python
# Count duplicates, remove them, and confirm the new shape (Tutorial 3).
#8 duplicates found; df.shape  (292, 6).
df.shape
```

```python
print(df.duplicated().sum()) # counts fully repeated rows
df = df.drop_duplicates()    # removes them. 
```

#### The new shape reveals 292, and recorded in the Cleaning Log.

```python
df.shape # 300 - 8 = 292
```

## Step 6: Repair the zip codes

#### Recall that zip was int64, we want to convert the zip to text or str.  This will restore the leading zeros.

```python
# Convert zip to text and restore the leading zeros using Tutorial 4 adapted to 5-character zips.
df["zip"] = df["zip"].astype(str).str.zfill(5) # zips like 02134 display with their leading zero.
```

```python
print(df.head(7)) # leading zeros are visible on "zip"
```

## Step 7: Standardize the dates

#### Previously, date was an str type.  The date can be converted to datetime64[]. The min and max can be obtained this way from the describe() function.

```python
# Convert the date column to real datetimes (Tutorial 5) and verify the dtype.
# datetime64[ns] in the type list.
df["date"] = pd.to_datetime(df["date"], format="mixed")
```

## Step 8: Investigate the negative quantities

#### The negative quantities could be faulty entries, and taking the absolute value retains the accuracy of the data.  They should be addressed by using the absolute value or by fixing them by hand.  However absolute value is more efficient.

```python
# Use the filtering pattern (Tutorial 6) to display rows where qty is below zero
# Decide: returns to flag, typos to correct, or rows to remove. Apply your choice in code.
df[df["qty"] < 0]
```

```python
# absolute values of each can be taken
# Use absolute value to correct negative quantities, from the math library
df["qty"] = df["qty"].abs()
```
