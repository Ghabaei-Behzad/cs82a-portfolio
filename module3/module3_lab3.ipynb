#### For "span" we can study the relationship with "lanes" to decide if span is: SHORT, MEDIUM, or LONG. (Problem 2.2 part c. i.)
#### It may be useful to flag these 16 entries with a new column called span_flag.  Here are the missing span data points.[27]: Filling the empty span data with MEDIUM is the best solution. [31]:  Lets also create flag columns for the columns that are strings and that have large amounts of missing data that we fill in ourselves.  [28]: [29]: [30]:

```python
df[df["span"].isna()]
```

#### Making a flag column for span called "span_flag".

```python
def span_condition(v):
    if v == "LONG" or v == "SHORT" or v == "MEDIUM":
        return "no flag"
    else:
        return "flag"
# the apply function was needed here
df['span_flag'] = df["span"].apply(span_condition)
df.tail(10)
```

#### Making a flag column for t-or-d called "t_or_d_flag".

```python
def t_or_d_condition(v):
    if v == "THROUGH" or v == "DECK":
        return "no flag"
    else:
        return "flag"
# the apply function was needed here
df['t_or_d_flag'] = df["t-or-d"].apply(t_or_d_condition)
df.head(10)
```

#### Making a flag column for rel-l called "rel_l_flag".

```python
def rel_l_condition(v):
    if v == "S" or v == "F" or v == "S-F":
        return "no flag"
    else:
        return "flag"
# the apply function was needed here
df['rel_l_flag'] = df["rel-l"].apply(rel_l_condition)
df.head(10)
```

#### A good solution for missing span values, is to fill the span column with MEDIUM.

# Fill in span

```python
# span is missing 16
fill_value = "MEDIUM" 
df["span"] = df["span"].fillna(fill_value)
```

#### We next look at "erected" dates in integer form, and we should fill the rel-l column with strings that were common at their time peiod. Thus, before the year 1884 as S and after 1884 with F for rel-l.

# Fill in rel-l

```python
# 1. Fill missing values with 'S' where 'erected' is less than 1884
df.loc[df["erected"] < 1884, "rel-l"] = df["rel-l"].fillna("S")

# 2. Fill the remaining missing values with 'F' (where 'erected' >= 1884)
df.loc[df["erected"] >= 1884, "rel-l"] = df["rel-l"].fillna("F")
```
